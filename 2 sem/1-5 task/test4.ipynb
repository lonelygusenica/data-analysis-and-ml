```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from MyKmeans import KMeans
from MyDbscan import MyDBSCAN
from sklearn.metrics import silhouette_score
```

```python
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=1.2, random_state=42)
```

```python
plt.scatter(X[:,0], X[:,1])
plt.title("Исходные данные")
plt.show()
```

```python
my_kmeans = KMeans(n_clusters=3, max_iter=20)
my_kmeans.fit(X)
```

```python
labels = my_kmeans.predict(X)
```

```python
plt.scatter(X[:,0], X[:,1], c=labels)
plt.scatter(my_kmeans.centroids[:,0], my_kmeans.centroids[:,1],
            color="red", marker="X", s=200)

plt.title("Мой KMeans")
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
sk_kmeans = KMeans(n_clusters=3, random_state=42)
sk_labels = sk_kmeans.fit_predict(X)
```

```python
plt.scatter(X[:,0], X[:,1], c=sk_labels)
plt.scatter(sk_kmeans.cluster_centers_[:,0],
            sk_kmeans.cluster_centers_[:,1],
            color="red", marker="X", s=200)
```

```python
silhouette_score(X, labels)
```

```python
silhouette_score(X, sk_labels)
```

```python
my_db = MyDBSCAN(eps=1.2, min_samples=5)
my_db.fit(X)
```

```python
plt.scatter(X[:,0], X[:,1], c=my_db.labels)
plt.title("Мой DBSCAN")
plt.show()
```

```python
from sklearn.cluster import DBSCAN
```

```python
sk_db = DBSCAN(eps=1.2, min_samples=5)
sk_labels = sk_db.fit_predict(X)
```

```python
plt.scatter(X[:,0], X[:,1], c=sk_labels)
plt.title("Sklearn DBSCAN")
plt.show()
```

```python
silhouette_score(X, labels)
```

```python
silhouette_score(X, sk_labels)
```
