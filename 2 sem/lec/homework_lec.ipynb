```python
import pandas as pd
import numpy as np
from sklearn.semi_supervised import LabelSpreading
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('text_embs.csv')
```

```python
df.head(5)
```

Ручная разметка

```python
df['target'] = -1
```

0 - космос, 1 - биология

```python
df.loc[1, 'target'] = 0
```

```python
df.loc[10, 'target'] = 0
```

```python
df.loc[4, 'target'] = 1
```

```python
df.loc[7, 'target'] = 1
```

```python
df.head(11)
```

```python
X = df[['e1', 'e2', 'e3', 'e4', 'e5']].values
y = df['target'].values
```

```python
lp_model = LabelSpreading(kernel='knn', n_neighbors=5)
lp_model.fit(X, y)
```

```python
predicted_labels = lp_model.transduction_
```

```python
label_map = {0: 'space', 1: 'biology'}
df['article_type'] = [label_map[label] for label in predicted_labels]
```

```python
df.head(11)
```

```python
df.to_csv('text_embs_labeled.csv', index=False)
```
