```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from MyKNN import MyKNN
from sklearn.neighbors import KNeighborsClassifier
```

```python
data = load_iris()
X = data.data
y = data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
knn = MyKNN(k=5)
knn.fit(X_train, y_train)
predictions = knn.predict(X_test)
print(f"my KNN: {knn.accuracy(X_test, y_test):.2f}%")
```

```python
knn_sklearn = KNeighborsClassifier(n_neighbors=5)
knn_sklearn.fit(X_train, y_train)
acc_sklearn = knn_sklearn.score(X_test, y_test) * 100
print(f"Точность KNN из sklearn: {acc_sklearn:.2f}%")
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic = titanic[['pclass', 'sex', 'age', 'sibsp', 'parch', 'fare', 'survived']].dropna()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
titanic['sex'] = le.fit_transform(titanic['sex'])
```

```python
X = titanic.drop('survived', axis=1).values
y = titanic['survived'].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
knn = MyKNN(k=5)
knn.fit(X_train, y_train)
predictions = knn.predict(X_test)
print(f"my KNN: {knn.accuracy(X_test, y_test):.2f}%")
```

```python
knn_sklearn = KNeighborsClassifier(n_neighbors=5)
knn_sklearn.fit(X_train, y_train)
acc_sklearn = knn_sklearn.score(X_test, y_test) * 100
print(f"Точность KNN из sklearn: {acc_sklearn:.2f}%")
```
